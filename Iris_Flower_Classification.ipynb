7. Load the Built-In Iris Dataset and Display Column Names

````python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df = iris.frame
print(df.columns)
````

8. Map Target Values to Flower Species and Visualize Sepal Measurements

````python
df['species'] = df['target'].map({
    0: 'Setosa',
    1: 'Versicolor',
    2: 'Virginica'
})


sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species'
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
````

9. Separate Input Features and Target Variable

````python
x = df.drop('species', axis=1)
y = df['species']
````

10. Split the Dataset into Training and Testing Sets

````python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
````

11. Standardize the Input Features

````python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

12. Initialize and Train the K-Nearest Neighbours (KNN) Model

````python
model=KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
````

13. Predict Flower Species on Test Data

````python
y_pred = model.predict(X_test)
````

14. Evaluate Model Accuracy

````python
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy}')
````

15. Generate Predictions for Precision Evaluation

````python
y_pred = model.predict(X_test)
````

````python
16. Calculate Model Precision
````

````python
precision = precision_score(y_test, y_pred, average='weighted')
print(f'Precision: {precision}')
````

17. Display the Confusion Matrix

````python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
````
